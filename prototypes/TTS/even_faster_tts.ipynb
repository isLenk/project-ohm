# Even Faster TTS!
using KoljaB/RealtimeTTS

```python
from queue import Queue
from RealtimeTTS import TextToAudioStream, CoquiEngine
import torch
```

```python
def check_cuda():
    """Simply display whether CUDA is available and print device information.
    CoquiEngine will automatically use CUDA if available."""

    if torch.cuda.is_available():
        print("CUDA available.")
        print(f"Current: {torch.cuda.current_device()} - Count: {torch.cuda.device_count()}")
        print(f"Device Name: {torch.cuda.get_device_name(torch.cuda.current_device())}")
    
```

```python
audio_queue = Queue()
chunks_received = 0
engine = None
engine_stream = None
pending_feed = Queue()

def load_engine(model_name="xtts_v2", voice="voices/lance.wav"):
    global engine, engine_stream, pending_feed, audio_queue, chunks_received
    check_cuda()
    audio_queue = Queue()
    chunks_received = 0
    engine = CoquiEngine(use_deepspeed=True)
    engine_stream = TextToAudioStream(engine)
    pending_feed = Queue() 
```

```python
load_engine()
```

```python
words = "Waddup hommie?"
while words != "exit":
    words = input("Enter text to synthesize (or 'exit' to quit): ")
    if words == "exit":
        break

    engine_stream.feed(words).play()

print("Synthesis complete. Exiting...")
```
