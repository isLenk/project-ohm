## For API Registry
Checks if API's are available.

```python
import aiohttp
import asyncio
```

```python
registry = {
    "frontend": "localhost:8000",
    "tts": "localhost:8001",
    "text-gen": "localhost:8002",
}
```

```python

async def get_health(url: str) -> str:
    async with aiohttp.ClientSession() as session:
        try:
            async with session.get(f"http://{url}/health") as response:
                return f"{url}: {response.status}"
        except (aiohttp.ClientError, asyncio.TimeoutError):
            return f"{url}: ERROR"
        except Exception as e:
            return f"{url}: {e}"
    
async def main():
    tasks = [get_health(url) for url in registry.values()]
    responses = await asyncio.gather(*tasks)

    print("RESULTS")
    for response in responses:
        print(response)

await main()
```
