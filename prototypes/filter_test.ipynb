# Filter Testing
Automatically bleep or kill depending on swear-to-message threshold

```python
filter_bleep = ["dumb", "brainless", "stupid"]
filter_kill = ["idiot", "suck"]
```

```python
test_messages = [
    "You are so dumb",
    "You are so brainless",
    "You are so STUPID!!!!",
    "dumb Dumb DUMB du.mb DUMB",
    "You are an idiot",
    "You are a suck",
]

def filter_message(message, filter_list, kill_filter):
    """
    Filters out words from the message based on the filter list.
    """
    for word in filter_list:
        message = message.lower().replace(word.upper(), "*" * len(word))
        message = message.lower().replace(word.lower(), "*" * len(word))

    for word in kill_filter:
        if word.lower() in message:
            message = "CENSORED."
            break
        
    return message
```

```python
for message in test_messages:
    print(filter_message(message, filter_bleep, filter_kill))
```
