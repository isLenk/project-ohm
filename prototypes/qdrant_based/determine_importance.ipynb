Notebook for determining whether a prompt contains a potential memory

```python
sentences = [
    "humans can be very good at recognizing faces, but they are not so good at recognizing other things.",
    "dragons are real, i like dragons.",
    "be quiet please",
    "tell me more about your day",
    "i am a dragon, i like to fly.",
    "the eiffel tower does not exist.",
]

not_memories = [2,3, 4]
```

```python
# Chat completion
import openai
import os
import json
import time
import random
```

```python
# Send request to 127.0.0.1:5000

def send(data):
    client = openai.OpenAI(api_key="0608da5d28eb10cea2914f3de0f3ddba", base_url="http://127.0.0.1:5000/v1")
    response = client.chat.completions.create(
        model="gpt-3.5-turbo",
        messages=[
            {"role": "system", "content": "Determine if the following sentence is either a memory/fact/preference or not. Return only either 'yes' or 'no' or else I will be fired."},
            {"role": "user", "content": data}
        ],
        # stream=True
    )
    return response

nos = []
for i, sentence in enumerate(sentences):
    if sentence in not_memories:
        continue
    print(f"--> Sending: {sentence}")
    answer = ""
    while answer != "yes" and answer != "no":
        res = send(sentence)
        answer = res.choices[0].message.content.lower()
        print(answer)
        if answer == "no":
            nos.append(i)

if nos == not_memories:
    print("All tests passed")
else:
    print("Some tests failed")
    print(f"Expected: {not_memories}, Got: {nos}")
```
